```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load Data 
data = pd.read_csv('plant_data.csv')


#encode soil type (
soil_dummies = pd.get_dummies(data['Soil Type'], prefix='Soil')
data = pd.concat([data, soil_dummies], axis=1)

# Encode crop type
le_crop = LabelEncoder()
crop_encoded = le_crop.fit_transform(data['Crop Type'])
data['Crop_Encoded'] = crop_encoded

# Now use one-hot encoded soil columns
feature_columns = ['Temparature', 'Moisture', 'Humidity']
X = data[feature_columns].values
y = data['Crop_Encoded'].values


#split
X_train, X_test, y_train, y_test =train_test_split(X,
                 y,
                 test_size = 0.20,
                 random_state= 45)
#scale
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

#model
knn = KNeighborsClassifier(n_neighbors = 70)
knn.fit(X_train, y_train)
print(knn.score(X_test_scaled,y_test ))

print("actual value", y[20])
print("predicted value", knn.predict(X)[20])oist
```
